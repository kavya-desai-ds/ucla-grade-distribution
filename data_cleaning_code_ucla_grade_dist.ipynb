```python
'''
import data
'''

import pandas as pd

url1 = "https://raw.githubusercontent.com/kavya-desai-ds/ucla-grade-distribution/61da81e6600f66f230b48b2e74c55b4a7e1fb24d/ucla%20grades%20fall21-summer22.csv"
url2 = "https://raw.githubusercontent.com/kavya-desai-ds/ucla-grade-distribution/61da81e6600f66f230b48b2e74c55b4a7e1fb24d/ucla%20grades%20fall22-spring23.csv"
url3 = "https://raw.githubusercontent.com/kavya-desai-ds/ucla-grade-distribution/61da81e6600f66f230b48b2e74c55b4a7e1fb24d/ucla%20grades%20summer23-spring24.csv"
url4 = "https://raw.githubusercontent.com/kavya-desai-ds/ucla-grade-distribution/61da81e6600f66f230b48b2e74c55b4a7e1fb24d/ucla%20grades%20summer24-spring25.csv"

rawdf1 = pd.read_csv(url1)
rawdf2 = pd.read_csv(url2)
rawdf3 = pd.read_csv(url3)
rawdf4 = pd.read_csv(url4)
```

```python
'''
clean df1
'''
#fix colnames
cleandf1 = rawdf1.drop(columns=['ENROL TERM SEQ NBR', 'INTL CATL', 'INTL SECT', 'CREATION DATE', 'ROWNO'])
new_col_names = ['Enrollment Term', 'Subject Area', 'Catalogue Number', 'Section Number', 'Letter Grade',
       'Letter Grade Count', 'Total Enrolled', 'Instructor Name', 'Instructor Code', 'Course Title']
cleandf1.columns = new_col_names

#remove classes not taught by a prof
cleandf1 = cleandf1.loc[cleandf1['Instructor Code'] == 1]
cleandf1 = cleandf1.drop(columns=['Instructor Code'])

#remove non-letter grades like P/NP, etc.
grades_to_keep = ['A+', 'A', 'A-', 'B+', 'B', 'B-', 'C+', 'C', 'C-', 'D+', 'D', 'D-', 'F']
cleandf1['Letter Grade'] = cleandf1['Letter Grade'].str.strip()
cleandf1 = cleandf1.loc[cleandf1['Letter Grade'].isin(grades_to_keep)]

#create new col with numeric gpa instead of letter grade
letter_to_gpa = {
    "A+" : 4.0,
    "A" : 4.0,
    "A-" : 3.7,
    "B+" : 3.3,
    "B" : 3.0,
    "B-" : 2.7,
    "C+" : 2.3,
    "C" : 2.0,
    "C-" : 1.7,
    "D+" : 1.3,
    "D" : 1.0,
    "D-" : 0.7,
    "F" : 0.0
}

cleandf1["Numeric Grade"] = cleandf1["Letter Grade"].map(lambda row: letter_to_gpa[row])

#enrollment term fix
enrollment_year = {
    "21F" : 2021,
    "22W" : 2022,
    "22S" : 2022,
    "221" : 2022,
    "222" : 2022
}
enrollment_q = {
    "21F" : "Fall",
    "22W" : "Winter",
    "22S" : "Spring",
    "221" : "Summer",
    "222" : "Summer"
}

cleandf1["Year"] = cleandf1["Enrollment Term"].map(lambda row: enrollment_year[row])
cleandf1["Quarter"] = cleandf1["Enrollment Term"].map(lambda row: enrollment_q[row])
cleandf1.head()
```
